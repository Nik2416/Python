**BINARY SEARCH**



```python
# ITERATIVE SOLUTION
def bin(nums,t):
   n=len(nums)
   low=0
   high=n-1
   while low<=high:
    mid=(low+high)//2
    if nums[mid]==t:
      return mid
    elif nums[mid]<t:
      low=mid+1
    else:
      high=mid-1
   return -1

bin([1,2,3,4,6,8,1],1)
```

```python
#Recursive solution
def b(nums,l,h,t):
  if l<h:
    mid=(l+h)//2
    if nums[mid]==t:
      return mid
    elif nums[mid]<t:
      return b(nums,mid+1,h,t)
    else:
      return b(nums,l,mid-1,t)

b([1,2,3,4,7,6,8],0,5,1)
```

```python
#BINARY SEARCH
#LEETCODE 35 -SEARCH INSERT POSITION
def searchPosition(nums,target):
  #return index of the target in O(logn)
  n=len(nums)
  lb=n
  l=0
  r=n-1
  while l<r:
    m=(l+r)//2
    if nums[m]>=target:
      lb=m
      r=m-1
    else:
      l=m+1
  return lb

searchPosition([1,3,4,5,8,9,14,15,19,22],1)
```

```python
#Ceil and Floor -----LOWER BOUND AND UPPER BOUND
#Find the Ceil and Floor of a number
def lb(nums,target):
  n=len(nums)
  l=0
  h=n-1
  f=-1
  c=-1
  while l<h:
    m=(l+h)//2
    if nums[m]==target:
      return [nums[m],nums[m]]
    elif nums[m]>target:
      c=nums[m]
      h=m-1
    else:
      f=nums[m]
      l=m+1
  return [f,c]

lb([1,3,4,5,8,9,14,15,19,22],1)
```
