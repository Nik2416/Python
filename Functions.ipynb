**What Are Functions?**

A block of code that takes an input and returns an output

Let's create a function with DocString

```python
def is_even(i):  #here i is parameter
  """
  This Function returns if the number is even or odd
  input - any valid integer
  output - even or odd
  """
  if type(i)== int:
    if i%2==0:
      return "Even"
    return "Odd"
  print('Pass an integer')
print(is_even(221)) #here argument is a parameter
```

```python
for i in range(1,11):
  print(is_even(i))
```

**Keyword Arguments**

*   *args
*   **kwargs

Both are special python keywords that are used to pass  the variable length of aarguments to a function

```python
#args allows us to pass a variable number of non keyword arguments to a function

def multiple(*args):
  product =1

  for i in args:
    product*=i

  return product
```

```python
multiple(1,2,3,4,5,6,7,8,9)
```

```python
#kwargs allows us to pass any number of keywords args
#key-value pair

def country(**kwargs):

  for (key,value) in kwargs.items():
    print(key,'->',value)
  print(kwargs)

country(America='Washington DC',India='Delhi ',Australia='Melbourne')
```

**Lambda Function**

A lambda function is a small anonymous function.
A lambda function can take any number of args but can only have one expression.




```python
a=lambda x:x**2
```

```python
a(2)
```

```python
b= lambda s:'a' in s
b('niyati')
```

```python
c= lambda x:"even" if  x%2==0 else 'odd'
```

```python
c(298)
```

```python
# def square(x):
#   return x**2

square=lambda x:x**2
#LAMBDA FUNCTIONS ARE USED WITH HOF
#HOF=  High Order Functions are functions where you takes a FUNCTION as an input


def transform(f,L):
  output=[]

  for i in L:
    output.append(f(i))
  return output

List=[1,2,3,4,5,6,7,8,9,12]
transform(square,List)
```

```python
#map function = maps an iterable to some given function

l=[1,2,3,4]
list(map(lambda x:x**2, l))
```

```python

list(map(lambda x:'even' if x%2==0 else 'odd' , l))
```

```python
users =[
    {
        'name':'NISHA',
        'age':45
    },
    {
        'name':'Nitish',
        'age':56
    }
]

list(map(lambda users:users['name'],users ))
list(map(lambda users:users['age'],users ))
```
