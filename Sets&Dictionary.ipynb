**SET**

```python
set1={1,2,3,3,4,5,6}
print(set1)
print(type(set1))
print(len(set1))
```

```python
st={1,2,4}
st2=st
print(st)
print(st2)
st.add(5)
print(st)
print(st2)
```

```python
st2=st.copy()
print(st2)
st.add(6)
print(st)
print(st2)
```

```python
set2={2,8,9}

#UNION
print(set1|set2)
print(set1.union(set2))

#INTERSECTION
print(set1&set2)
print(set1.intersection(set2))

#DIFFERENCE
print(set1-set2)
print(set2.difference(set1))

#SYMMETRIC DIFFERENCE
print(set1^set2)
print(set2.symmetric_difference(set1))
```
