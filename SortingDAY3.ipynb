**Selection sort**

```python
def selection(arr):
  for i in range(0,len(arr)):
     min=i
     for j in range(len(arr)):
       if arr[min]>arr[j]:
        min=j
        arr[i],arr[min]=arr[min],arr[i]
  return arr

n=[5,7,8,4,1,6,9,2,3]
selection(n)
```

```python
def selectiona(arr):
  for i in range(0,len(arr)):
     min=i
     for j in range(i+1,len(arr)):
       if arr[j]<arr[min]:
        min=j
     arr[i],arr[min]=arr[min],arr[i]
  return arr

n=[5,7,8,4,1,6,9,3,2]
selectiona(n)
```

**Bubble sort**

```python
def bubble(arr):
  n=len(arr)
  for i in range(n-2,-1,-1):
    for j in range(0,i+1):
      if arr[j]>arr[j+1]:
       arr[j],arr[j+1]=arr[j+1],arr[j]
  return arr

n=[5,7,8,4,1,6,9,3,2]
bubble(n)
```

```python
def maximum(n):
  max=n[0]
  for i in range(len(n)):
      if max < n[i]:
        max=n[i]
  return max
```

```python
#array is sorted or not
def nums(n):
  for i in range(len(n)-1):
      if n[i]>n[i+1]:
         return False
  return True

nums([1,11,21,31,34,45,55])
```

```python
#returning how many new elements are there in an array
def re(n):
  k={}
  for i in range(len(n)):
      k[n[i]]=0

  l=0
  for j in k:
      n[l]=j
      l+=1
  return l,k

re([1,1,2,2,2,212,11,21,31,34,45,55])
```

```python
#remove dulicates
def du(nums):
    i = 0
    j = i + 1

    while j < len(nums):
        if nums[i] == nums[j]:
            pass
        else:
            i += 1
            nums[i], nums[j] = nums[j], nums[i]
        j += 1
    return nums, i + 1

du([1,1,2,2,2,212,11,21,31,34,45,55])
```

```python
#Right Rotate by 1

def rotate(nums):
  n=len(nums)
  nums[:] = [nums[n-1]] + nums[0:n-1]
  return nums

n=[45,5,6,7,8]
rotate(n)
```

```python
# rotate by 1 solution
def ro(nums):
  tem=nums[-1]
  for i in range(len(nums)-2,-1,-1):
    nums[i+1]=nums[i]
  nums[0]=tem
  return nums

n=[45,5,6,7,8]
ro(n)
```

```python
# rotate by k solution
def ro(nums,k):
  n=len(nums)
  for _ in range(0,k):
      if k>n:
        k=k%n
      else:
        tem=nums.pop()
        nums.insert(0,tem)
  return nums

n=[45,5,6,7,8]
ro(n,k=3)
```

```python
#optimal rotate solution LEETCODE=189
def rot(nums,k):
  n=len(nums)
  k=k%n
  nums[:]=nums[n-k:]+nums[:n-k]
  return nums
n=[45,5,6,7,8]
rot(n,k=3)
```

```python
#Reverse
def reverse(nums,left ,right):
  while left<right:
    nums[left],nums[right]=nums[right],nums[left]
    left+=1
    right-=1
  return nums

nums=[9,23,34,11,5]
reverse(nums,0,2)
```
