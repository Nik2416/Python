```python
#OPTIMAL SET MATRIX ZEROS
#LEETCODE 73
def zero(matrix):
  r=len(matrix)
  c=len(matrix[0])
  rt=[0]*r
  ct=[0]*c
  for i in range(0,r):
     for j in range(0,c):
        if matrix[i][j]==0:
           rt[i]=-1
           ct[j]=-1

  for i in range(0,r):
     for j in range(0,c):
        if rt[i]==-1 or ct[j]==-1:
          matrix[i][j]=0
  return matrix

nums=[[7,9,5,2],[10,22,0,1],[29,0,8,5],[4,14,6,7]]
zero(nums)

'''
FIRST WE TOOK NUMBER OF ROWS AND COLUMNS
THEN WE WE TAKE ROW TRACK LIST AND COLUMNTRACK LIST
TO TRACK IF THERES A ZERO IN
THAN WE UPDATE OUR TRACK LIST TO -1
AND THEN WE ITERATE AGAIN AND
UPDATE OUR MATRIX ELEMENT TO ZERO
WHERE THE TRACK LIST IS -1
'''
```

```python
#Leetcode 48: Rotate Matrix by 90 Degrees
def rotate(matrix):
  n=len(matrix)
  c=len(matrix[0])

  for i in range(0,n-1):
    for j in range(i+1,n):
      matrix[i][j],matrix[j][i]=matrix[j][i],matrix[i][j]
  for i in range(0,n):
    matrix[i].reverse()
  return matrix

nums=[[7,9,5,2],[10,22,0,1],[29,0,8,5],[4,14,6,7]]
rotate(nums)
```
