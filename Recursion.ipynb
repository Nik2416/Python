```python
#To print x ,n times

def func(x,n):
  if n==0:
    return
  print(x)
  func(x,n-1)

func(15,4)
```

```python
#To print 1 to n
def num(i,n):
  if i>n:
    return
  num(i+1,n)
  print(i)
num(1,5)
```

```python
def sum(n):
  return n*(n+1)//2

sum(910)
```

```python
#parametrised recursion
def rsum(sum=0,n=1):
  if n==0:
    print(sum)
    return
  sum+=n
  rsum(sum,n-1)

rsum(0,910)
```

```python
#Functional Recursion-- where we rutrn something ,we dont print
#Create a flow
#create the base condition

def fun(n):
  if n==1:
    return 1
  return n+fun(n-1)
fun(5)
```

```python
#fact of a number
def fact(n):
  if n==1:
     return 1
  return n*fact(n-1)

fact(5)
fact(4)
fact(12)
fact(10)
```

```python
#Reverse an array using recursion
def reverse(num,l,r):
  if l==r:
    return num
  if l>r:
    return num
  tem=num[l]
  num[l]=num[r]
  num[r]=tem
  return reverse(num,l+1,r-1)

n=[1,2,3,4]
reverse(n,1,3)

```

```python
#check if it is a palindrome

def palin(s,l,r):
  if l==r:
    return 0
  elif l>r:
    return 1
  if s[l]==s[r]:
    return True
  else:
    return False
  palin(s,l+1,r-1)


```

```python
s='121'
st='anbcddcbna'
palin(s,0,2)
palin(st,0,9)
```
