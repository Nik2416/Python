**Intro to Linked List**



```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None

a=Node(5)
b=Node(6)
c=Node(7)

a.next=b
b.next=c

head=a

print(head.data)
print(head.next.data)
```

```python
#Traversing in a linked list
def traverseLL(head):
  curr=head
  while curr !=None:
    print(curr.data)
    curr=curr.next
```

**INSERTION**

```python
#INSERTION in the Beginning

newnode=Node(4)
newnode.next=head
head=newnode

traverseLL(head)
```

```python
#Insertion at the end
newNode=Node(78)
curr=head

while curr.next!=None:
   curr=curr.next

curr.next=newNode
traverseLL(head)
```

```python
#insertion at kth index
k=2
nn=Node(8)
curr =head

for i in range(k-1):
  curr=curr.next

nn.next=curr.next
curr.next=nn

traverseLL(head)
```

**DELETION IN LL**

```python
#deletion at the first index
head=head.next
traverseLL(head)
```

```python
#deletion at the last index
curr=head

while curr.next.next!=None:
   curr=curr.next

curr.next=None

traverseLL(head)
```

```python
#delete in the middle
k=2

curr=head
for i in range(k-1):
  curr=curr.next

curr.next=curr.next.next
traverseLL(head)
```
