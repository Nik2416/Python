**Characteristics of a list**


*   Ordered
*   Changeable
*   Heterogenous
*   Dynamic
*  Can be nested
*   Have duplicates
*   Items can be accessed
*   can contain any kind of objects in python














```python
#HOW TO CREATE A LIST
#empty
print([])
```

```python
#1d list
print([1,2,3,4,5])
```

```python
#2d list
print([1,2,3,4,[1,2,3,4]])
```

```python
#3d list
print([[[1,2],[3,4]],[[5,6],[7,8]]])
```

```python
#Using type conversion
print(list('hello'))
```

**Accessing Items from a list**

```python
#Indexing
L=[1,2,3,4,5]
#positive indexing
print(L[0])
#Negative Indexing
print(L[-2])

l=[2,3,[3,4]]
print(l)
print(l[2][0])
print(l[-1][1])
#slicing

print(l[1:])
```

ADDING AN ELEMENT TO A LIST

```python
#append - adds a single element to list
l.append(7)
l

#extend - adds multiple elements to a list
l.extend([5,77])
l
```

```python
L.append([5,6,7])
L.extend('hello')
L
```

```python
#insert =(index, what to add)

b=[1,2,5]
b.insert(1,144)
```

```python
b
```

Deleting items from a list

```python
#del
del l[1]
l
```

```python
#remove -when u only know the number and not the index
print(l)
l.remove(77)
print(l)
```

```python
#pop -is used to delete last item of the list
print(l)
l.pop()
#also u can provide the index
l.pop(2)
print(l)
```

```python
#clear
print(l)
l.clear()
print(l)
```
