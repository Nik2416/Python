```python
#Leetcode 2149: Rearrange Array Elements by Sign
def rearrnge(nums):
  pos=[]
  neg=[]
  n=len(nums)
  for i in range(0,n):
    if nums[i]>0:
       pos.append(nums[i])
    else:
       neg.append(nums[i])
  for i in range(0,len(pos)):
      nums[2*i]=pos[i]
      nums[2*i+1]=neg[i]
  return nums

```

```python
def orearrange(nums):
  n=len(nums)
  pos=0
  neg=1
  r=[0]*n
  for i in range(0,n):
    if nums[i]>=0:
      r[pos]=nums[i]
      pos+=2
    else:
      r[neg]=nums[i]
      neg+=2
  return r

orearrange([5,10,-1,-3,6,-10])
```

```python
#Longest Consecutive Substring
#Leetcode 128
def con(nums):
    m = 0
    n = len(nums)

    for i in range(0, n):
        nu = nums[i]
        count = 1

        while nu + 1 in nums:
            count += 1
            nu += 1

        m = max(m, count)   # ← this must be inside FOR

    return m
nums=[1,2,44,98,99,100,101]
con(nums)
```

```python
#LEETCODE 905. Sort Array By Parity
def sortArrayByParity(nums):
    n=len(nums)
    start=0
    for i in range(0,n):
      if nums[i]%2==0:
        nums[start],nums[i]=nums[i],nums[start]
        start+=1
    return nums

sortArrayByParity([3,1,2,4])
```

```python
#128: longest consecutive
def longestConsecutive(nums):
  n=len(nums)
  mset=set()
  longest=0
  for i in range(n):
    mset.add(nums[i])
  for num in mset:
    if num-1 not in mset:
      x=num
      count=1
      while x+1 in mset:
        x+=1
        count+=1
      longest=max(longest,count)
  return longest

longestConsecutive([100,4,200,1,3,2])
```
